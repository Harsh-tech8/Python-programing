## ABSTRACT METHOD

```python
from abc import ABC , abstractmethod
```

```python
class payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
```

```python
class upi(payment):
    def pay (self, amount):
        print("paid RS.",amount,"using upi")
        
```

```python
class credit_card(payment):
    def pay(self , amount):
         print("paid RS.",amount,"using ccredit card")
        
```

```python
UPI = upi()
```

```python
UPI.pay(14000)
```

```python
cc = credit_card()
```

```python
cc.pay(4000)
```
