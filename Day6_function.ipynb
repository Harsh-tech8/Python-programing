```python
def cheak_len(password):
    return len(password) >= 8
```

```python
cheak_len("harsh1233")
```

```python
def has_lowercase(password):
    for char in password:
        if char.islower():
            return True
    return False       
```

```python
def has_uppercase(password):
    for char in password:
        if char.isupper():
            return True
    return False       
```

```python
has_uppercase("AADDVfg")
```

```python
def has_digit(password):
    for char in password:
        if char.isdigit():
            return True
    return False 
```

```python
has_digit("harsh12345")
```

```python
def has_special (password):
    special ="!@#$%^&*"

    for char in password:
        if char in special:
            return True

    return False        
```

```python
def security_score(password):
    score = 0

    if cheak_len(password):
        score += 1  
    if has_digit(password):
        score += 1
    if has_special(password):
        score += 1    
    if has_uppercase(password):
        score += 1
    if has_uppercase(password):
        score += 1

    return score
        



        

```

```python
def password_levelcheak(score):
    if score >= 4:
        print (f"Very storge password the sore is:{score} ")   
    elif score >= 3:
        print (f"Very medime password the sore is:{score} ")
    elif score >= 2:
        print (f"Very imporve password the sore is:{score} ")    
    elif score >= 1:
        print (f"Very weak password the sore is:{score} ")
    else :
        print("type another password")
    
```

```python
def password_report(score, password):
    print("=" * 30)
    print("   PASSWORD SECURITY REPORT   ")
    print("=" * 30)
    print(f"Password Tested: {password}\n")
    
    print("Criteria Checklist:")
    print(f"- Length check {score}/5:      {'[PASS]' if cheak_len(password) else '[FAIL]'}")
    print(f"- Contains digit {score}/5:    {'[PASS]' if has_digit(password) else '[FAIL]'}")
    print(f"- Contains special {score}/5:  {'[PASS]' if has_special(password) else '[FAIL]'}")
    print(f"- Contains uppercase {score}/5: {'[PASS]' if has_uppercase(password) else '[FAIL]'}")
    
    print("\nSummary:")
    password_levelcheak(score)
    print("=" * 30)
    
```

```python
while True:
    print("\n" + "=" * 25)
    print("        MAIN MENU        ")
    print("=" * 25)
    print("1. Create the report")
    print("2. Exit")
    print("=" * 25)

    choice = input("Enter your choice (1 or 2): ").strip()

    if choice == "1":
        
        password = input("Enter the lab password: ")
        
       
        score = security_score(password)
        password_report(score, password)
        
    elif choice == "2":
        print("Exiting the program. Goodbye!")
        break
    else:
        print("Invalid choice! Please enter 1 or 2.")
```
