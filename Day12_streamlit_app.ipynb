```python
import streamlit as st

st.title("OUR FIRST APP")
st.write("Welocome to our stramlit app")

name = st.text_input("enter you name")

python_mark = st.number_input("Python mark",min_value = 0, max_value=100)
linux_mark = st.number_input("linux mark",min_value = 0, max_value=100)
cloud_mark = st.number_input("cloud mark",min_value = 0, max_value=100)

if st.bottom("Calculate Result "):
    totalmark = python_mark+linux_mark+cloud_mark
    per = totalmark / 3

    if per >= 90:
        grade = "A+"
    elif per >= 80:
        grade = "A"
    elif per >= 70:
        grade = "B" 
    elif per >= 60:
        grade = "B+"
    elif per >= 50:
        grade = "C"
     else :
         grade = "Fail"

    st.write("Name of student :",name)
    st.write("persantage:",round(per,2),)
    st.write("Total Mark:",totalmark)
    st.success("Grade Mark:",grade)

    "sub":["Python","Linux","Cloud"]
    "Marks":[python_mark,linux_mark,cloud_mark]
```
