```python
def total_exp():
    total = 0
    with open("exp.txt","r") as file:
        line = file.readlines()
        for line in line :
           data =line.strip().split(",")
           amount=float(data[1])
           total = total + amount
        print("total exp is :" , total )
```

```python
total_exp()
```

```python
def higtest_amount():
    hightest = 0
    hightest_cat = ""

    with open ("exp.txt","r") as file:
        line = file.readlines()
        for line in line:
            category,amount = line.strip().split(",")
            amount = float(amount)
            hightest_cat = category
        print("higtest exp is :",hightest_cat) 
        print("amount",hightest)
```

```python
higtest_amount()
```
