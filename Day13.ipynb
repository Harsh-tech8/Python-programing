```python
import tkinter as tk

window = tk.Tk()
window.title("My first TK app")
window.geometry("500x400")

heading = tk.Label(window, text="MY LW APP", font=("Arial", 24))
heading.pack(pady=30)

name_label = tk.Label(window, text="enter your name:", font=("Arial", 14), width=25)
name_label.pack()

name_enter = tk.Entry(window, font=("Arial", 14), width=25)
name_enter.pack(pady=10)

def welcome():
    name = name_enter.get()
    result_label.config(text=f"Welcome {name}")

welcome_button = tk.Button(window, text="Welcome ME", font=("Arial", 14), command=welcome)
welcome_button.pack(pady=10)

result_label = tk.Label(window, text="", font=("Arial", 14))
result_label.pack(pady=20)

window.mainloop()
```
