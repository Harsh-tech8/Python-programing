```python
"""
08 - Search Google / YouTube using Python

YouTube:
    pip install google-api-python-client

Google:
For automated Google web search, prefer an official Google Programmable
Search / Custom Search API or another permitted search API. Do not build
automation that bypasses Google's anti-bot systems.

The YouTube example below uses the official YouTube Data API.
"""
```

```python
from googleapiclient.discovery import build


YOUTUBE_API_KEY = "PUT_YOUR_YOUTUBE_API_KEY_HERE"


def youtube_search(query, max_results=5):
    youtube = build(
        "youtube",
        "v3",
        developerKey=YOUTUBE_API_KEY,
    )

    request = youtube.search().list(
        part="snippet",
        q=query,
        type="video",
        maxResults=max_results,
    )

    response = request.execute()

    results = []
    for item in response.get("items", []):
        video_id = item["id"]["videoId"]
        title = item["snippet"]["title"]

        results.append({
            "title": title,
            "url": f"https://www.youtube.com/watch?v={video_id}",
        })

    return results


if __name__ == "__main__":
    query = input("Search YouTube for: ")

    for result in youtube_search(query):
        print("\n", result["title"])
        print(result["url"])
```
