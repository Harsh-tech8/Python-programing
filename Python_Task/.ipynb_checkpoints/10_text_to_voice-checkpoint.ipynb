# 10 Text To Voice



```python
"""
10 - Text to voice using Python

Install:
    pip install pyttsx3

pyttsx3 normally works offline and uses the speech engine available on your OS.
"""
```

```python
import pyttsx3


def text_to_voice(text):
    engine = pyttsx3.init()

    engine.setProperty("rate", 160)
    engine.setProperty("volume", 1.0)

    engine.say(text)
    engine.runAndWait()


if __name__ == "__main__":
    text = input("Enter text to speak: ")
    text_to_voice(text)
```
