```python
"""
01 - Send messages: Email, SMS, and WhatsApp

Install:
    pip install twilio pywhatkit

EMAIL:
Uses Gmail SMTP. For Gmail, use an App Password rather than your normal password.

SMS / WHATSAPP:
Twilio is used for both. You need a Twilio account and verified/approved numbers
according to Twilio's current requirements.

Note: pywhatkit's WhatsApp automation opens WhatsApp Web and is suitable for
personal/simple automation. For production systems, prefer an official API.
"""

import smtplib
from email.message import EmailMessage

# ---------- EMAIL ----------
def send_email(sender, app_password, receiver, subject, body):
    msg = EmailMessage()
    msg["From"] = sender
    msg["To"] = receiver
    msg["Subject"] = subject
    msg.set_content(body)

    with smtplib.SMTP_SSL("smtp.gmail.com", 465) as server:
        server.login(sender, app_password)
        server.send_message(msg)

# Example:
# send_email(
#     "yourgmail@gmail.com",
#     "YOUR_GMAIL_APP_PASSWORD",
#     "receiver@example.com",
#     "Test Email",
#     "Hello from Python!"
# )


# ---------- SMS / WHATSAPP USING TWILIO ----------
def send_twilio_sms(account_sid, auth_token, from_number, to_number, message):
    from twilio.rest import Client

    client = Client(account_sid, auth_token)
    result = client.messages.create(
        body=message,
        from_=from_number,
        to=to_number,
    )
    return result.sid


def send_twilio_whatsapp(account_sid, auth_token, from_whatsapp, to_whatsapp, message):
    from twilio.rest import Client

    client = Client(account_sid, auth_token)
    result = client.messages.create(
        body=message,
        from_=from_whatsapp,  # e.g. "whatsapp:+14155238886" for a Twilio setup
        to=to_whatsapp,       # e.g. "whatsapp:+91XXXXXXXXXX"
    )
    return result.sid


# ---------- SIMPLE WHATSAPP WEB OPTION ----------
def send_whatsapp_web(phone, message, hour, minute):
    import pywhatkit

    # Opens WhatsApp Web and sends at the specified local time.
    pywhatkit.sendwhatmsg(phone, message, hour, minute)


if __name__ == "__main__":
    print("Edit the examples above with your own credentials before running.")
```
